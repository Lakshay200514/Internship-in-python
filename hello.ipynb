```python
ls=[]
while(True):
    print("""
    1. Add element
    2. Remove element
    3. Display first three element
    4. Display last three element
    5. Print
    6. Exit
    """)
    ch=int(input("Enter your choice:"))
    if ch==1:
        e=input("Enter the element:")
        ls.append(e)
    elif ch==2:
        if len(ls) ==0 :
            print("Empty List")
        else:
            c=int(input("Enter the element:"))
            ls.remove(ls[c])
    elif ch==3:
        print(ls[0:3])
    elif ch==4:
        print(ls[-3:])
    elif ch==5:
        print(ls)
    elif ch==6:
        break
    else:
        print("Invalid Choice")
        break
print("bye")
```
