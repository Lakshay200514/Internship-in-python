```python
import pandas as pd
import numpy as np
```

```python
df=pd.DataFrame({
    'col1':range(12),
    'col2':['A']*3+['B']*3+['C']*3+['D']*3,
    'date':pd.to_datetime(['2021-01-01','2021-01-02','2021-01-03']*4)
})
print(df)
pivoted=df.pivot(index='date',columns='col2',values='col1')
print(pivoted)
```

```python
data={
    'course':['Cs','civil','Cs','civil','Cs'],
    'year':[1,2,1,1,2],
    'Student':[100,150,120,190,100]
}
df=pd.DataFrame(data)
df.loc[5]=['Python',1,200]
df['Marks']=[90,80,85,70,95,100]
df['Marks']=df['Student'][:3]
print(df)
pivoted=df.pivot_table(index='course',columns='year',values='Student',aggfunc='sum')
print(pivoted)
```

```python
data1={
    'col1':[2,np.nan,np.nan,4],
    'col2':[1,pd.NA,pd.NA,3]
}
df=pd.DataFrame(data1)
f=df.fillna('-')
d=df.ffill()
e=df.bfill()
print(d)
print(f)
print(e)
```

```python
df=pd.DataFrame([[9,3,4],[4,5,6],[7,8,9]],index=['a','b','c'],columns=['one','two','three'])
print(df)
df=df.reindex(['a','b','c','d','e'])
f=df.ffill(limit=1)
print(df)
print(f)
```

```python
data2={
    'col1':[1,2,3],
    'col2':[5,6,7],
    'col3':[9,10,11]
}
df=pd.DataFrame(data2)
print(df.replace({5:50,9:90}))
```

```python
data3={
    'one':['A','B','C','D','A','B'],
    'two':[1,2,3,4,5,6],
}
df=pd.DataFrame(data3)
f=df.groupby('one')['two'].sum()
g=df.groupby('one')['two'].mean()
h=df.groupby('one')['two'].median()
i=df.groupby('one')['two'].tail(1)
j=df.groupby('one')['two'].max()
k=df.groupby('one')['two'].min()
print(f)
print(g)
print(h)
print(i)
print(j)
print(k)
```

```python
data3={
    'one':['A','B','C','D','A','B'],
    'subCategotry':['X','Y','Z','X','Y','Z'],
    'two':[1,2,3,4,5,6],
}
df=pd.DataFrame(data3)
print(df)
multiple_group=df.groupby(['one','subCategotry'])['two'].sum()
print(multiple_group)
```
