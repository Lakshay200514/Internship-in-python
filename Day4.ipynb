```python
def maximum(a,b,c):
    if a>b and a>c:
        print(f"{a} is max")
    elif b>a and b>c:
        print(f"{b} is max")
    else:
        print(f"{c} is max")
x=int(input("Enter the number"))
y=int(input("Enter the number"))
z=int(input("Enter the number"))
maximum(x,y,z)
```

```python
def distinct_list(ls):
    ls=set(ls)
    ls=list(ls)
    print(ls)
li=[]
k=int(input("Enter the range"))
for i in range(k):
    l=int(input("Enter the number"))
    li.append(l)
distinct_list(li)
```

```python
def multiplication_list(lst,num):
    lst=list(lst)
    for j in range(len(lst)):
        lst[j] *= num
li=[]
k=int(input("Enter the range"))
for i in range(k):
    l=int(input("Enter the number"))
    li.append(l)
no=int(input("Enter the number to multiply"))
multiplication_list(li,no)
print(li)
```

```python
def factorial(num):
    num=int(num)
    if num==0 or num==1:
        return 1
    else:
        return num*factorial(num-1)
n=int(input("Enter the number"))
factorial(n)
```

```python
def reverse_string(string):
    string=string[::-1]
    return string
a=input("Enter the string")
reverse_string(a)
```

```python
def even_list(ls):
    ls=list(ls)
    for j in ls:
        if j%2==0:
            print(j)
li=[1,2,3,4,5,6]
even_list(li)
```

```python
def prime_number(number):
    if number==0 or number==1:
        return False

    for j in range(2,number):
        if number%j==0:
            return False
    else:
        return True
num = int(input("Enter a number: "))

if prime_number(num):
    print("Prime")
else:
    print("Not Prime")
```

```python
def check_range(number, start, end):
    if start <= number <= end:
        return True
    else:
        return False


num = int(input("Enter number: "))
s = int(input("Enter start range: "))
e = int(input("Enter end range: "))

if check_range(num, s, e):
    print("Number is within range")
else:
    print("Number is outside the range")
```

```python
def string(str1):
    count=0
    lower=0
    for m in str1:
        if m.isupper():
            count+=1
        elif m.islower():
            lower+=1

    print(f"Uppercase letters: {count}")
    print(f"Lowercase letters:  {lower}")
s = input("Enter a string: ")
string(s)
```
