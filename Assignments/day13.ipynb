```python
arr=np.array([[23, 56, 78, 93],[71, 82,13, 24]])
for i in range(len(arr)):
    for j in range(len(arr[i])):
        if arr[i][j]%2 != 0:
            arr[i][j]=-1
print(arr)
            
```

```python
arr=np.array([1, 0, 2, 0, 3, 0, 4])
for i in range(len(arr)):
    if(arr[i]!=0):
        print(i)
res=arr.nonzero()
print(res)
```
