```python
import pandas as pd
```

```python
df=pd.DataFrame(
    {
        "date":["2025-01-01","2026-04-02","2024-12-03"],
        "value":[10,20,30]
    }
)
print(df)
df["date"]=pd.to_datetime(df["date"])
print(df)
df["year"]=df["date"].dt.year
df["month"]=df["date"].dt.month 
df["day"]=df["date"].dt.day
print(df)
df['Month'] = df['date'].dt.month_name()
print(df)
```
