```python
arr=np.array([[6, -8, 73, -110], [np.nan, -8, 0, 94]])
print(arr)
```

```python
col_mean=np.nanmean(arr,axis=0)
rows,col=np.where(np.isnan(arr))
arr[rows,col]=col_mean
```
