```python
a=input("Enter your name:")
b=input("Enter your class:")
arr=[]
for i in range(5):
    c=int(input("Enter your marks:"))
    arr.append(c)
total=0
for i in range(5):
    total+=arr[i]
percentage=(total/500)*100
print("result:")
print(a)
print(b)
for i in range(5):
    print(arr[i])
print(percentage)
if percentage>=60:
    print("A")
elif percentage>=50:
    print("B")
elif percentage>=40:
    print("C")
elif percentage>=33:
    print("D")
else:
    print("Fail")
```
