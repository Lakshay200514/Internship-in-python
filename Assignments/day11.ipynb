```python
arr = np.array([1, -2, 3, -4, 5])
arr[arr < 0] = 0
print(arr)
```

```python
arr = np.array([
    [1, 2, np.nan],
    [4, np.nan, 6],
    [7, 8, 9]
])
print(arr)
col_mean = np.nanmean(arr, axis=0)
inds = np.where(np.isnan(arr))
arr[inds] = np.take(col_mean, inds[1])
print(arr)
```
