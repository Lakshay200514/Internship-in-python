```python
import pandas as pd
df=pd.DataFrame({'A':[1,2,3,4,5],'B':[10,20,30,40,50]})
print(df+2)
print(df-2)
print(df*2)
print(df/2)
print(df[1:4]+2)
print(df)
```

```python
df=pd.DataFrame({'A':[1,2,3,4,5],'B':[10,20,30,40,50]})
df1=pd.DataFrame({'A':[1,2,3],'B':[10,20,30]},index=[1,2,3])
print(df)
print(df1)
print(df+df1)
```

```python
#dataFrame cncatenation
df1=pd.DataFrame({'A':[1,2,3],'B':[10,20,30]})
df2=pd.DataFrame({'A':[4,5,6],'B':[40,50 ,60]})
df3=pd.concat([df1,df2],ignore_index=False,axis=1)
print(df3)
```

```python
one=pd.DataFrame({
    "Name":['Alice','Bob','Charlie','Dave'],
    "Subject":['Math','Science','History','English'],
    "Marks":[85,90,78,92]
},index=[1,2,3,4])
two=pd.DataFrame({
    "Name":['Eve','Frank','Grace','Heidi'],
    "Subject":['Math','Science','History','English'],
    "Marks":[98,96,98,99]
},index=[1,2,3,4])
res=pd.concat([one,two],keys=['x','y'])
res1=pd.concat([one,two],ignore_index=True)
res3=pd.concat([one,two],ignore_index=True,axis=1)
res2=pd.concat([one,two],ignore_index=True)
print(res)
print(res1)
print(res2)
print(res3)
```

```python
left=pd.DataFrame({
    "id":[1,2,3,4],
    "Name":['Alice','Bob','Charlie','Dave'],
    "Subject":['Math','Science','History','English']
})
right=pd.DataFrame({
    "id":[1,2,3],
    "Name":['Charlie','Dave','Eve'],
    "Subject":['History','English','Math']
})
res=pd.merge(left,right,on='id',how='inner')
res1=pd.merge(left,right,on='id',how='outer')
res2=left.merge(right,on=['id'],how='left')
res3=left.merge(right,on=['id'],how='right')
# res4=left.merge(right,on=['id'],rsuffixes=('_left','_right'))
res4=left.join(right,lsuffix='Left',rsuffix='Right')
print(res)
print(res1)
print(res2)
print(res3)
print(res4)
```
