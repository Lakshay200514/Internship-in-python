```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='sin(x)')
plt.title('Graph of sin(x)')
plt.xlabel('x axis')
plt.ylabel('Sin(x)')

plt.legend()
plt.grid(True)
plt.show()
```

```python
categories=['A','B','C','D']
values=[10,20,30,40]
plt.bar(categories,values,color='lightblue')
plt.title("Box PLot")
plt.xlabel("categories")
plt.ylabel("Values")
plt.show()
```

```python
data=np.random.rand(1000)
plt.hist(data,bins=100,alpha=0.9)
plt.title("Histogram")
plt.xlabel("xaxis")
plt.ylabel("y axis")
plt.show()
```
