```python
arr2=np.array([[1,2,3,4,5],[7,8,9,5,6],[11,12,13,21,14]])
print(arr2)
arr2[[0,2],:]=arr2[[2,0],:]
arr2[:,[2,3,4]]=arr2[:,[4,3,2]]
print(arr2)
```

```python
arr3=np.array(
    [
    [
        [4,5,6],
        [7,9,5]
    ],
    [
        [8,9,1],
        [7,4,1]
    ]
    ]
)
print(arr3)
arr3[:,[0,1],:]=arr3[:,[1,0],:]
print(arr3)
arr3[[0,1],[0,1],:]=arr3[[1,0],[1,0],:]
print(arr3)
```

```python
arr4=np.array([1,2,np.nan,5,np.nan])
isnan=np.isnan(arr4)
print(isnan)
arr5=arr4[~np.isnan(arr4)]
print(arr5)
res=np.nan_to_num(arr4,copy=True,nan=0)
print(res)
```

```python
arr1=np.array([1,2,3])
arr2=np.array([[1,2,3],[7,8,9]])
np.savez('data2.npz',a=arr1,b=arr2)
res=np.load('data2.npz')
print(res['a'])
print(res['b'])
```

```python
arr1=np.array([[-3,2],[1,-2]]) # 3x+2y and x+2y
arr2=np.array([5,5]) # these are constant so the full equation is 3x+2y=5 and x+2y=5
res=np.linalg.solve(arr1,arr2)
print(res)
```

```python
arr1_inv=np.linalg.inv(arr1)
print(arr1_inv)
res=np.dot(arr1_inv,arr2)
print(res)
```
