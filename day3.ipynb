```python
a=int(input("Enter the number"))
b=int(input("Enter the number"))
while(True):
    print("""choose any option for basic math operation:
    1. Addition
    2. Subtraction
    3. Multiplication
    4. Division""")
    ch=int(input("Enter your choice"))
    if ch==1:
        print(a+b)
        break
    elif ch==2:
        print(a-b)
        break
    elif ch==3:
        print(a*b)
        break
    elif ch==4:
        if b<a:
            print(a/b)
        else:
            print(b/a)
            break
    else:
        print("Invalid choice")
        break
```
